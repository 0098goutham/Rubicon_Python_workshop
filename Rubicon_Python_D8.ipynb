```python
import pandas as pd
data=pd.read_csv('zepto_sales_dataset.csv')
data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x='City', data=data)
plt.title('Number of Orders by City')
plt.xlabel('City')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figure=(3,5))

plt.subplot(3,4,1)
sns.histplot(data['Original Price'], kde=True, bins=30, color='skyblue')
sns.kdeplot(data["Original Price"], color='red')
plt.title('DIstribution of Original price', fontsize=16)
plt.xlabel('Original Price', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(True)#import matplotlib.pyplot as plt
import seaborn as sns
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20,12))
plt.subplot(3,4,1)
sns.histplot(data['Original Price'],kde=True,bins=30,color='skyblue')
sns.kdeplot(data['Original Price'],color='red') #kde line in red
plt.title('Distribution of Original Price',fontsize=16)
plt.xlabel('Original Price',fontsize=12)
plt.ylabel('Frequency',fontsize=12)
plt.grid(True)

#Display the plot
plt.tight_layout()
plt.show()
```

```python
avg_discount_by_category = data.groupby('Category')['Discount'].mean().reset_index()
plt.figure(figsize=(20,12))
# 2. Bar plot of average discount by category
plt.subplot(4,4,2)
sns.barplot(x='Category' , y='Discount', data=avg_discount_by_category, palette='viridis')

plt.title('Average discount by category',fontsize=16)
plt.xlabel('Category',fontsize=12)
plt.ylabel('Average Discount',fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(True)

#Display the plot
plt.tight_layout()
plt.show()
```

```python
data_sorted = data.sort_values(by='Category')
plt.figure(figsize=(12, 8))
sns.lineplot(x='Category', y='Total Revenue', data=data_sorted, marker='v', line_style='-', markersize=12)
plt.title('Total Revenue by Category', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)
plt.xticks(rotation=45)
# plt.grid(True)
# plt.tight_layout()
plt.show()
```
